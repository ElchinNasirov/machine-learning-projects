```python
# Neural Network Backpropagation

import numpy as np

print("Neural Network Backpropagation\n")

# Simple example with one hidden layer

# Training data (XOR-like)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

y = np.array([[0],
              [1],
              [1],
              [0]])

# Network parameters
np.random.seed(42)
W1 = np.random.randn(2, 4) * 0.5   # Input → Hidden
b1 = np.zeros((1, 4))
W2 = np.random.randn(4, 1) * 0.5   # Hidden → Output
b2 = np.zeros((1, 1))

learning_rate = 0.1
epochs = 5000

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_deriv(a):
    return a * (1 - a)

# Training loop
for epoch in range(epochs):
    # Forward Propagation
    Z1 = np.dot(X, W1) + b1
    A1 = sigmoid(Z1)

    Z2 = np.dot(A1, W2) + b2
    A2 = sigmoid(Z2)

    # Cost
    cost = np.mean((A2 - y)**2)

    # Backpropagation
    dZ2 = (A2 - y) * sigmoid_deriv(A2)
    dW2 = np.dot(A1.T, dZ2)
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * sigmoid_deriv(A1)
    dW1 = np.dot(X.T, dZ1)
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    # Update weights
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    if epoch % 1000 == 0:
        print(f"Epoch {epoch} | Cost: {cost:.4f}")

# Final predictions
print("\nFinal Predictions:")
print(A2.round(3))
```
