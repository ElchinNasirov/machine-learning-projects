```python
# Bias vs Variance + Learning Curves

import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression

# Generate sample regression data
X, y = make_regression(n_samples=200, n_features=1, noise=20, random_state=42)

# Learning Curve
train_sizes, train_scores, val_scores = learning_curve(
    LinearRegression(),
    X, y,
    cv=5,
    scoring='neg_mean_squared_error',
    train_sizes=np.linspace(0.1, 1.0, 10)
)

# Convert to positive MSE
train_scores_mean = -train_scores.mean(axis=1)
val_scores_mean = -val_scores.mean(axis=1)

# Plot Learning Curve
plt.figure(figsize=(10, 6))
plt.plot(train_sizes, train_scores_mean, 'o-', color='blue', label='Training Error')
plt.plot(train_sizes, val_scores_mean, 'o-', color='red', label='Validation Error')
plt.xlabel('Training Set Size')
plt.ylabel('Mean Squared Error')
plt.title('Learning Curve (Bias vs Variance Diagnosis)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("Interpretation:")
print("- High Bias (Underfitting): Both training and validation error are high")
print("- High Variance (Overfitting): Training error is low, validation error is high")
print("- Good Fit: Both errors are low and close to each other")

print("\n completed!!!")
```
