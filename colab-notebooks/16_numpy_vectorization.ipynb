```python
#Vectorization with NumPy; Optimize Code

import numpy as np
import time


# 1. Non-vectorized vs Vectorized Addition
size = 1000000
a = list(range(size))
b = list(range(size))

# Slow way (Python loop)
start = time.time()
c = [a[i] + b[i] for i in range(size)]
loop_time = time.time() - start
print(f"Python loop time: {loop_time:.4f} seconds")

# Fast way (NumPy vectorization)
a_np = np.arange(size)
b_np = np.arange(size)

start = time.time()
c_np = a_np + b_np
numpy_time = time.time() - start
print(f"NumPy vectorized time: {numpy_time:.4f} seconds")
print(f"NumPy is approximately {loop_time / numpy_time:.0f}x faster\n")

# 2. Vectorized operations examples
print("Vectorized Operations Examples:")
arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("arr * 2     :", arr * 2)
print("arr ** 2    :", arr ** 2)
print("arr > 3     :", arr > 3)
print("Mean        :", np.mean(arr))
print("Sum         :", np.sum(arr))

print("\nDone!")
```
