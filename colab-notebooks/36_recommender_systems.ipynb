```python
# Recommender Systems (Content-Based)

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity


# Simple movie feature matrix
# Features: Action, Comedy, Romance
movies = {
    "Movie A": [1, 0, 1],
    "Movie B": [1, 1, 0],
    "Movie C": [0, 1, 1],
    "Movie D": [1, 0, 0],
    "Movie E": [0, 1, 0]
}

movie_names = list(movies.keys())
X = np.array(list(movies.values()))

# User likes Movie A
user_movie = np.array([movies["Movie A"]])

# Similarity between user movie and all movies
similarity = cosine_similarity(user_movie, X)[0]

print("Recommendations based on 'Movie A':")
ranked = sorted(zip(movie_names, similarity), key=lambda x: x[1], reverse=True)

for name, score in ranked:
    print(f"{name}: {score:.2f}")

print("\nKey Concepts:")
print("- Content-based filtering recommends items similar to what the user already likes")
print("- Cosine similarity measures how close two items are")
print("- Collaborative filtering uses other users' ratings instead of item features")
```
