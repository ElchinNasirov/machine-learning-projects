```python
#Neural Networks – Forward Propagation

import numpy as np

print("Neural Network Forward Propagation\n")

# Simple 2-layer neural network (from scratch)

# Input (2 features)
X = np.array([[1.0, 2.0]])          # Shape: (1, 2)

# Layer 1 weights and bias
W1 = np.array([[0.5, -0.3],         # Shape: (2, 2)
               [0.8,  0.4]])
b1 = np.array([[0.1, 0.2]])         # Shape: (1, 2)

# Layer 2 weights and bias
W2 = np.array([[0.6],               # Shape: (2, 1)
               [-0.5]])
b2 = np.array([[0.3]])              # Shape: (1, 1)

# ----- Forward Propagation -----

# Layer 1
Z1 = np.dot(X, W1) + b1             # Linear transformation
A1 = np.maximum(0, Z1)              # ReLU activation

print("Layer 1 (Z1):", Z1)
print("Layer 1 (A1 - after ReLU):", A1)

# Layer 2
Z2 = np.dot(A1, W2) + b2            # Linear transformation
A2 = 1 / (1 + np.exp(-Z2))          # Sigmoid activation

print("\nLayer 2 (Z2):", Z2)
print("Layer 2 (A2 - Final Output):", A2)
```
